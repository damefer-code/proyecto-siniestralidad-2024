```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv("../data/processed/accidentes_2024_limpio.csv")
df.shape
```

```python
tabla_chi_via = pd.crosstab(
    df["TIPO_VIA_DESC"],
    df["ACCIDENTE_GRAVE"]
)

tabla_chi_via
```

```python
chi2, p, dof, expected = chi2_contingency(tabla_chi_via)

chi2, p, dof
```

```python
n = tabla_chi_via.to_numpy().sum()

phi2 = chi2 / n
r, k = tabla_chi_via.shape

v_cramer = np.sqrt(
    phi2 / min(k - 1, r - 1)
)

v_cramer
```

```python
El test Chi-cuadrado indica una asociación estadísticamente significativa entre el tipo de vía y la gravedad del accidente (χ² = 397,13; p < 0,001).

Sin embargo, la V de Cramér obtenida es 0,088, lo que indica que la intensidad de la asociación es débil.
```
