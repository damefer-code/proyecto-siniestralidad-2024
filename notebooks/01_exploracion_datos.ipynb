```python
df_ine = pd.read_csv(archivo_ine, sep=";", encoding="latin1")
```

```python
df_ine.head()
```

```python
df_ine_provincias = df_ine[df_ine["Provincias"].notna()].copy()
```

```python
df_ine_provincias.shape
```

```python
df_ine_provincias["Provincias"].tolist()
```

```python
df_ine_provincias["COD_PROVINCIA"] = (
    df_ine_provincias["Provincias"]
    .str[:2]
    .astype(int)
)
```

```python
df_ine_provincias[["Provincias", "COD_PROVINCIA"]].head(10)
```

```python
df_ine_provincias["POBLACION_2024"] = (
    df_ine_provincias["Total"]
    .str.replace(".", "", regex=False)
    .astype(int)
)
```

```python
df_ine_provincias[["Provincias", "Total", "POBLACION_2024"]].head(10)
```

```python
df_ine_final = df_ine_provincias[
    ["COD_PROVINCIA", "Provincias", "POBLACION_2024"]
].copy()
```

```python
df_ine_final.head()
```

```python
df_ine_final["COD_PROVINCIA"].duplicated().sum()
```

```python
df_final = df_seleccion.merge(
    df_ine_final[["COD_PROVINCIA", "POBLACION_2024"]],
    on="COD_PROVINCIA",
    how="left",
    validate="many_to_one"
)
```
