```python
df["DIA_SEMANA"].value_counts().sort_index()
```

```python
mapa_dias = {
    1: "Lunes",
    2: "Martes",
    3: "Miércoles",
    4: "Jueves",
    5: "Viernes",
    6: "Sábado",
    7: "Domingo"
}

df["DIA_SEMANA_DESC"] = df["DIA_SEMANA"].map(mapa_dias)
```

```python
df["FIN_SEMANA"] = df["DIA_SEMANA"].isin([6, 7]).astype(int)
```

```python
df["FIN_SEMANA"].value_counts()
```

```python
def clasificar_franja(hora):
    if 0 <= hora < 6:
        return "Madrugada"
    elif 6 <= hora < 12:
        return "Mañana"
    elif 12 <= hora < 18:
        return "Tarde"
    else:
        return "Noche"

df["FRANJA_HORARIA"] = df["HORA"].apply(clasificar_franja)
```

```python
df["FRANJA_HORARIA"].value_counts()
```

```python
df["ACCIDENTE_GRAVE"] = (
    (df["TOTAL_MU24H"] > 0) |
    (df["TOTAL_HG24H"] > 0)
).astype(int)
```

```python
df["ACCIDENTE_GRAVE"].value_counts()
```

```python
df["ACCIDENTE_MORTAL"] = (
    df["TOTAL_MU24H"] > 0
).astype(int)
```

```python
df["ACCIDENTE_MORTAL"].value_counts()
```

```python
df["TIENE_HERIDOS_GRAVES"] = (
    df["TOTAL_HG24H"] > 0
).astype(int)
```

```python
df["TIENE_HERIDOS_GRAVES"].value_counts()
```
